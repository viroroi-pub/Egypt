# Edge-Ramp Monte Carlo Generator
Generate per-phase Monte Carlo trials for the integrated edge-ramp model.
Input: friction (mu) and slope angle (theta_deg). Output: CSV.

```
from edge_ramp_montecarlo import save_trials_csv
mu, theta, N = 0.15, 7.0, 10000
out = '/mnt/data/montecarlo_edge_ramp_trials_mu0p15_theta7_10000.csv'
save_trials_csv(mu, theta, N, out)
out
```

```
# Example for another scenario
from edge_ramp_montecarlo import save_trials_csv
save_trials_csv(mu=0.30, theta_deg=8.0, N=10000, out='/mnt/data/montecarlo_edge_ramp_trials_mu0p30_theta8_10000.csv')
```

**Output columns**: trial_id, mu, theta_deg, phase_id, phase_label, ramp_configuration, ramps_up, ramps_down, turns_per_block, speed_factor, headway_arrival_min, corner_delay_per_turn_min, service_time_min, arrival_rate_blocks_per_min, service_rate_blocks_per_min, utilization, per_ramp_capacity_blocks_per_hour, total_capacity_blocks_per_hour, mean_corner_queue_length, waiting_probability, blocking_probability, settings, S_mu_theta.
